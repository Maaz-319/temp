```python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END
```

```python
class AgentState(TypedDict):
    question: str
    answer: str
```

```python
def take_input(state: AgentState) -> AgentState:
    state["answer"] = input("Q: Enter 'answer': ")
    return state
```

```python
def evaluator(state: AgentState):
    return "take_input" if state["answer"] != "answer" else "theEND"
```

```python
def theEND(state: AgentState):
    print("BYE")
    return state
```

```python
graph = StateGraph(AgentState)

graph.add_node("inp", take_input)
graph.add_node("theEND", theEND)    
graph.add_conditional_edges(
    "inp",
    evaluator,
    {
        "take_input": "inp",
        "theEND": "theEND"
    }
)
graph.set_entry_point("inp")
graph.add_edge("theEND", END)
app = graph.compile()
```

```python
results = app.invoke({})
```
